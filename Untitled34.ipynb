```python
import os
import json
import pickle
import warnings

import numpy as np
import pandas as pd
import yaml

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping


warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# ================================================================
# STEP 2: DEFINE FILE PATHS
# ================================================================

DATASET_PATH = r"C:\Users\sagni\Downloads\AI-Based Air Pollution Level Prediction\3b01bcb8-0b14-4abf-b6f2-c1bfd384ba69.csv"

OUTPUT_FOLDER = r"C:\Users\sagni\Downloads\AI-Based Air Pollution Level Prediction"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)


H5_PATH = os.path.join(
    OUTPUT_FOLDER,
    "air_pollution_model.h5"
)

PKL_PATH = os.path.join(
    OUTPUT_FOLDER,
    "air_pollution_model.pkl"
)

YAML_PATH = os.path.join(
    OUTPUT_FOLDER,
    "config.yaml"
)

JSON_PATH = os.path.join(
    OUTPUT_FOLDER,
    "metadata.json"
)


# ================================================================
# STEP 3: LOAD DATASET
# ================================================================

print("=" * 70)
print("AI-BASED AIR POLLUTION LEVEL PREDICTION")
print("=" * 70)

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )


df = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully.")
print("Dataset shape:", df.shape)

print("\nDataset columns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())


# ================================================================
# STEP 4: CLEAN COLUMN NAMES
# ================================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)


print("\nCleaned column names:")
print(df.columns.tolist())


# ================================================================
# STEP 5: CHECK REQUIRED TARGET COLUMN
# ================================================================

TARGET = "pollutant_avg"

if TARGET not in df.columns:
    raise ValueError(
        f"\nRequired target column '{TARGET}' was not found.\n"
        f"Available columns:\n{df.columns.tolist()}"
    )


# ================================================================
# STEP 6: CONVERT POLLUTION COLUMNS TO NUMERIC
# ================================================================

pollution_columns = [
    "pollutant_min",
    "pollutant_max",
    "pollutant_avg"
]


for column in pollution_columns:

    if column in df.columns:

        # Convert values such as NA, -, blank, etc. into NaN
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )


# Geographic columns should also be numeric

for column in ["latitude", "longitude"]:

    if column in df.columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )


# ================================================================
# STEP 7: REMOVE ROWS WITHOUT TARGET
# ================================================================

original_rows = len(df)

df = df.dropna(
    subset=[TARGET]
).copy()

removed_rows = original_rows - len(df)

print("\nRows removed because pollutant_avg was missing:",
      removed_rows)

print("Remaining rows:", len(df))


if len(df) < 10:
    raise ValueError(
        "Too few valid rows remain after cleaning the target."
    )


# ================================================================
# STEP 8: REMOVE OBVIOUS NON-PREDICTIVE COLUMNS
# ================================================================

# These columns may contain identifiers or update timestamps.
# We do not need them for the basic prediction model.

possible_drop_columns = [
    "id",
    "last_update",
    "country"
]


drop_columns = [
    column
    for column in possible_drop_columns
    if column in df.columns
]


# ================================================================
# STEP 9: CREATE FEATURE MATRIX AND TARGET
# ================================================================

X = df.drop(
    columns=[TARGET] + drop_columns,
    errors="ignore"
)

y = df[TARGET].astype(float)


# ================================================================
# STEP 10: REMOVE COMPLETELY EMPTY FEATURES
# ================================================================

all_null_columns = [
    column
    for column in X.columns
    if X[column].isna().all()
]


if all_null_columns:

    print("\nRemoving completely empty columns:")
    print(all_null_columns)

    X = X.drop(
        columns=all_null_columns
    )


# ================================================================
# STEP 11: IDENTIFY NUMERIC AND CATEGORICAL FEATURES
# ================================================================

numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()


categorical_features = X.select_dtypes(
    exclude=np.number
).columns.tolist()


print("\nNumeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


if len(numeric_features) + len(categorical_features) == 0:
    raise ValueError(
        "No usable input features were found."
    )


# ================================================================
# STEP 12: CREATE PREPROCESSING PIPELINE
# ================================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


transformers = []


if numeric_features:

    transformers.append(
        (
            "numeric",
            numeric_transformer,
            numeric_features
        )
    )


if categorical_features:

    transformers.append(
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    )


preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)


# ================================================================
# STEP 13: TRAIN-TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# ================================================================
# STEP 14: RANDOM FOREST MODEL
# ================================================================

random_forest_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=300,
                max_depth=None,
                min_samples_split=2,
                min_samples_leaf=1,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


print("\nTraining Random Forest model...")


random_forest_pipeline.fit(
    X_train,
    y_train
)


rf_predictions = random_forest_pipeline.predict(
    X_test
)


# ================================================================
# STEP 15: RANDOM FOREST PERFORMANCE
# ================================================================

rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)


rf_mse = mean_squared_error(
    y_test,
    rf_predictions
)


rf_rmse = np.sqrt(
    rf_mse
)


rf_r2 = r2_score(
    y_test,
    rf_predictions
)


print("\n" + "=" * 70)
print("RANDOM FOREST RESULTS")
print("=" * 70)

print("MAE :", round(rf_mae, 4))
print("MSE :", round(rf_mse, 4))
print("RMSE:", round(rf_rmse, 4))
print("R2  :", round(rf_r2, 4))


# ================================================================
# STEP 16: PREPROCESS DATA FOR ANN
# ================================================================

# Fit preprocessing only on training data

X_train_processed = preprocessor.fit_transform(
    X_train
)


X_test_processed = preprocessor.transform(
    X_test
)


# Convert to float32 for TensorFlow

X_train_processed = np.asarray(
    X_train_processed
).astype("float32")


X_test_processed = np.asarray(
    X_test_processed
).astype("float32")


y_train_ann = np.asarray(
    y_train
).astype("float32")


y_test_ann = np.asarray(
    y_test
).astype("float32")


print(
    "\nANN input feature count:",
    X_train_processed.shape[1]
)


# ================================================================
# STEP 17: BUILD ANN MODEL
# ================================================================

ann_model = Sequential(
    [
        tf.keras.Input(
            shape=(X_train_processed.shape[1],)
        ),

        Dense(
            128,
            activation="relu"
        ),

        Dropout(
            0.20
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dropout(
            0.10
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            1,
            activation="linear"
        )
    ]
)


ann_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)


print("\nANN Model Summary:")

ann_model.summary()


# ================================================================
# STEP 18: EARLY STOPPING
# ================================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)


# ================================================================
# STEP 19: TRAIN ANN
# ================================================================

print("\nTraining ANN model...")


history = ann_model.fit(
    X_train_processed,
    y_train_ann,

    validation_split=0.20,

    epochs=200,

    batch_size=32,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# ================================================================
# STEP 20: ANN PREDICTIONS
# ================================================================

ann_predictions = ann_model.predict(
    X_test_processed,
    verbose=0
).flatten()


# ================================================================
# STEP 21: ANN PERFORMANCE
# ================================================================

ann_mae = mean_absolute_error(
    y_test_ann,
    ann_predictions
)


ann_mse = mean_squared_error(
    y_test_ann,
    ann_predictions
)


ann_rmse = np.sqrt(
    ann_mse
)


ann_r2 = r2_score(
    y_test_ann,
    ann_predictions
)


print("\n" + "=" * 70)
print("ANN RESULTS")
print("=" * 70)

print("MAE :", round(ann_mae, 4))
print("MSE :", round(ann_mse, 4))
print("RMSE:", round(ann_rmse, 4))
print("R2  :", round(ann_r2, 4))


# ================================================================
# STEP 22: SAVE .H5 MODEL
# ================================================================

ann_model.save(
    H5_PATH
)


print(
    "\nH5 model saved successfully:"
)

print(H5_PATH)


# ================================================================
# STEP 23: SAVE .PKL FILE
# ================================================================

# The PKL contains:
# - Random Forest pipeline
# - fitted preprocessing information
# - input columns
# - target name

pkl_data = {

    "random_forest_pipeline":
        random_forest_pipeline,

    "ann_preprocessor":
        preprocessor,

    "feature_columns":
        X.columns.tolist(),

    "numeric_features":
        numeric_features,

    "categorical_features":
        categorical_features,

    "target":
        TARGET
}


with open(
    PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        pkl_data,
        file
    )


print(
    "\nPKL file saved successfully:"
)

print(PKL_PATH)


# ================================================================
# STEP 24: CREATE YAML CONFIGURATION
# ================================================================

config = {

    "project": {
        "name":
            "AI-Based Air Pollution Level Prediction",

        "task":
            "Regression",

        "target":
            TARGET
    },

    "dataset": {
        "path":
            DATASET_PATH,

        "original_rows":
            int(original_rows),

        "usable_rows":
            int(len(df)),

        "columns":
            int(df.shape[1])
    },

    "training": {
        "test_size":
            0.20,

        "random_state":
            42,

        "ann_epochs":
            200,

        "ann_batch_size":
            32
    },

    "random_forest": {
        "n_estimators":
            300,

        "mae":
            float(rf_mae),

        "rmse":
            float(rf_rmse),

        "r2":
            float(rf_r2)
    },

    "ann": {
        "mae":
            float(ann_mae),

        "rmse":
            float(ann_rmse),

        "r2":
            float(ann_r2)
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print(
    "\nYAML configuration saved successfully:"
)

print(YAML_PATH)


# ================================================================
# STEP 25: CREATE JSON METADATA
# ================================================================

metadata = {

    "project_name":
        "AI-Based Air Pollution Level Prediction",

    "problem_type":
        "Regression",

    "target_variable":
        TARGET,

    "dataset_file":
        os.path.basename(DATASET_PATH),

    "dataset_shape": {
        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1])
    },

    "input_features":
        X.columns.tolist(),

    "numeric_features":
        numeric_features,

    "categorical_features":
        categorical_features,

    "models": [
        "Random Forest Regressor",
        "Artificial Neural Network"
    ],

    "random_forest_metrics": {

        "MAE":
            float(rf_mae),

        "MSE":
            float(rf_mse),

        "RMSE":
            float(rf_rmse),

        "R2":
            float(rf_r2)
    },

    "ann_metrics": {

        "MAE":
            float(ann_mae),

        "MSE":
            float(ann_mse),

        "RMSE":
            float(ann_rmse),

        "R2":
            float(ann_r2)
    },

    "saved_files": {

        "h5":
            os.path.basename(H5_PATH),

        "pkl":
            os.path.basename(PKL_PATH),

        "yaml":
            os.path.basename(YAML_PATH),

        "json":
            os.path.basename(JSON_PATH)
    }
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )


print(
    "\nJSON metadata saved successfully:"
)

print(JSON_PATH)


# ================================================================
# STEP 26: FINAL MODEL COMPARISON
# ================================================================

print("\n" + "=" * 70)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)


comparison = pd.DataFrame(
    {
        "Model": [
            "Random Forest",
            "ANN"
        ],

        "MAE": [
            rf_mae,
            ann_mae
        ],

        "RMSE": [
            rf_rmse,
            ann_rmse
        ],

        "R2": [
            rf_r2,
            ann_r2
        ]
    }
)


print(
    comparison.round(4).to_string(
        index=False
    )
)


# ================================================================
# STEP 27: DISPLAY GENERATED FILES
# ================================================================

print("\n" + "=" * 70)
print("GENERATED FILES")
print("=" * 70)


generated_files = [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH
]


for file_path in generated_files:

    if os.path.exists(file_path):

        file_size = os.path.getsize(
            file_path
        )

        print(
            f"[SAVED] {os.path.basename(file_path)} "
            f"({file_size / 1024:.2f} KB)"
        )

    else:

        print(
            f"[NOT FOUND] {file_path}"
        )


print("\n" + "=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    "\nAll files have been saved in:"
)

print(
    OUTPUT_FOLDER
)
```
